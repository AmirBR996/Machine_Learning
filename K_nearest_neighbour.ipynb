First, we import the `kagglehub` library to easily download datasets from Kaggle. We then download the 'breast-cancer-dataset' and store its local path in the `path` variable. Finally, we print the path to confirm the download location.

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("yasserh/breast-cancer-dataset")

print("Path to dataset files:", path)
```

We use the `os` module to list the contents of the downloaded dataset directory, which helps us identify the CSV file containing the data.

```python
import os
os.listdir(path)
```

Here, we import the `pandas` library, commonly used for data manipulation. We then read the `breast-cancer.csv` file into a pandas DataFrame called `df`.

```python
import pandas as pd
df = pd.read_csv(f"{path}/breast-cancer.csv")
```

To get a quick overview of the data, we display a random sample of 10 rows from the DataFrame `df`.

```python
df.sample(10)
```

We separate the features (independent variables) from the target variable (dependent variable). `X` will contain all columns except 'diagnosis' and 'id', and `y` will contain only the 'diagnosis' column.

```python
X = df.drop(['diagnosis', 'id'] , axis= 1)
y = df['diagnosis']
```

We import `StandardScaler` from `sklearn.preprocessing` to standardize our features. Standardization scales the data to have a mean of 0 and a standard deviation of 1, which is often beneficial for machine learning algorithms.

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

We use `LabelEncoder` from `sklearn.preprocessing` to convert the categorical target variable 'diagnosis' into numerical labels (0 and 1).

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()
y = encoder.fit_transform(y)
```

We split the dataset into training and testing sets using `train_test_split` from `sklearn.model_selection`. 80% of the data is used for training, and 20% for testing. `random_state=True` ensures reproducibility of the split.

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(X , y , train_size = 0.8 , random_state = True)
```

We import `KNeighborsClassifier` from `sklearn.neighbors` and initialize a k-Nearest Neighbors model. This model classifies data points based on the majority class among their `k` nearest neighbors.

```python
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors = 3)
```

We train the `KNeighborsClassifier` model using the training data (`X_train` and `y_train`).

```python
model.fit(X_train , y_train)
```

After training, we use the model to make predictions on the unseen test data (`X_test`), storing the predicted labels in `y_pred`.

```python
y_pred = model.predict(X_test)
```

Finally, we import `f1_score` and `accuracy_score` from `sklearn.metrics` to evaluate the performance of our model. We calculate and print both the accuracy and the F1-score of the model on the test set.

```python
from sklearn.metrics import f1_score , accuracy_score
accuracy_score = accuracy_score(y_test , y_pred)
f1score = f1_score(y_test , y_pred)
print(f"accuracy : {accuracy_score} , f1 score : {f1score}")
```
